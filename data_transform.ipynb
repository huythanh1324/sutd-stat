## Modelling table (one row per hour)

`features.build_model_table` turns the raw tables into one row per forecast time **t** (hour-ending, Singapore time). The target `y` is the island-mean PM2.5 at **t + 24 h**. Every input uses only data time-stamped at or before t, so nothing from the future leaks in. The function checks that `y` really is PM2.5 24 hours later.

Filters and joins:
- **PM2.5:** island mean of the 5 regions, only for hours where all 5 report. The table is reindexed to a full hourly grid, so missing hours stay as explicit NaN.
- **Weather:** only the 12 stations that report all 5 variables. Aggregated per station first (mean, or sum for rainfall), then averaged across stations. Hour-ending labels via `ceil("h")`, the same as `to_hourly` in `data.ipynb`.
- **Fires:** `VIIRS_SNPP_SP` only (one consistent sensor), `type == 0` (vegetation fires), low-confidence pixels (`l`) dropped. Each fire is assigned to a sector by its compass bearing from Singapore. Fires before 15 Sep were not downloaded, so the first 24 h (48 h for the lag) of fire windows are NaN rather than wrongly zero.

```python
from features import build_model_table

H = 24                                   # forecast horizon in hours
table = build_model_table(d, h=H)
table.head()
```

### Data dictionary: modelling table

| column | unit | definition |
|---|---|---|
| index `timestamp` | SGT | Forecast time t (hour-ending). |
| `target_time` | SGT | t + 24 h, the hour being predicted. |
| `pm25_mean` | µg/m³ | Mean of the 5 regional 1-h PM2.5 values at t. NaN if any region is missing. |
| `y` | µg/m³ | **Target:** `pm25_mean` at t + 24 h. |
| `pm25_lag0`, `pm25_lag24` | µg/m³ | `pm25_mean` at t and at t − 24 h (persistence inputs). |
| `hour_sin`, `hour_cos` | – | sin/cos(2π·hour/24): the daily cycle without a jump at midnight. |
| `is_weekend` | 0/1 | Saturday or Sunday. |
| `temp`, `rh` | °C, % | Hourly island mean of the 12 full stations. |
| `rain_1h`, `rain_24h` | mm | Rain in the hour ending at t, and summed over the last 24 h. |
| `ws_ms` | m/s | Mean wind speed (knots × 0.514). |
| `u`, `v` | m/s | Wind vector components: u > 0 blowing towards the east, v > 0 towards the north. |
| `frp_{S}_24h` | log(1+MW) | log(1 + summed fire radiative power) of fires in sector S detected in (t − 24 h, t]. Sectors are 90° wide, centred on N/E/S/W as seen from Singapore. |
| `frp_{S}_lag1d` | log(1+MW) | Same, over (t − 48 h, t − 24 h]. |
| `wind_from_{S}` | m/s | Wind speed blowing from sector S towards Singapore: max(0, −(u·sinβ + v·cosβ)). |
| `transport_{S}` | – | `frp_{S}_24h × wind_from_{S}`: fires that the wind is carrying towards Singapore. |
